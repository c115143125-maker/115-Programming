# Loop
loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
for i in range(10):
    print(i)
```

```python
for i in range(10):
    print(i+1)
```

```python
sum=0
for i in range(9):
    sum=sum+(i+1)
print(sum)
```

```python
for i in range(6):
    print("&" * (i+1))
```

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
```

```python
for i in range(10):
    print(f'1*{(i+1)}={1*(i+1)}')
```

```python
for j in range(1,6):
    print(f'1*{j}={1*j}', end='  ')
print()

i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}', end='  ')
print()   

for j in range(1,6):
    print(f'3*{j}={3*j}', end='  ')
print()
```

## 練習印出質數 ##
質數:一個數字N，除了自己本身和1以外，沒有任何數可以除盡，N就是質數。


```python
n=273
i=2
if 273%i==0:
    print(f"{273}不是質數")
i=3
if 273%i==0:
    print(f"{273}不是質數")
```

```python
for i in range(2,273):
    if 273 % i==0:
        print(f'{273}不是質數,因為可以被{i}除盡')
```

```python
n=273
isPrime=True#假設n是質數
for i in range(2,n):
    if n % i==0:
        print(f'{n}不是質數,因為可以被{i}除盡')
        isPrime=False#確定n不是質數
        break #只要數字能被除盡，就跳離迴圈
if isPrime == True:#判斷我們的假設是否還成立
    print(f'{n}是質數') 
```
