# Loop
loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
for i in range(10):
    print(i)
```

```python
for i in range(10):
    print(i+1)
```

```python
sum=0
for i in range(9):
    sum=sum+(i+1)
print(sum)
```

```python
for i in range(6):
    print("&" * (i+1))
```

```python
print(f'1*1={1*1}')
print(f'1*2={1*2}')
print(f'1*3={1*3}')
```

```python
for i in range(10):
    print(f'1*{(i+1)}={1*(i+1)}')
```

```python
print(list(range(5)))
```

```python
print(list(range(2,101,2)))
```

```python
for j in range(1,6):
    print(f'1*{j}={1*j}', end='  ')
print()

i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}', end='  ')
print()   

for j in range(1,6):
    print(f'3*{j}={3*j}', end='  ')
print()
```

```python
for i in range(1,6):
    for j in range(1,6):
        print(f'{i}*{j}={i*j}', end=' ')
        print()     
```
